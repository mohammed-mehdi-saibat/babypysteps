```python
import pandas as pd
from IPython.display import display

# 1. The Historical Data (E-commerce visitors)
data = {
    'minutes_on_site': [1.5, 2.0, 5.5, 8.0, 9.5, 12.0],
    'pages_visited':   [1,   2,   4,   5,   7,   8],
    'bought_product':  [0,   0,   0,   1,   1,   1]  # 0 = No, 1 = Yes
}
df_class = pd.DataFrame(data)

# 2. Splitting into X (Features) and y (Target)
X = df_class[['minutes_on_site', 'pages_visited']] # The behavior we track
y = df_class['bought_product'] # The final action we want to predict

print('--- X (The clues) ---')
display(X)

print('\n--- y (The target) ---')
display(y)
```

```python
from sklearn.linear_model import LogisticRegression

# 1. Initialize the algorithm
classifier = LogisticRegression()

# 2. Train the model
classifier.fit(X, y)

# 3. Create two brand new, unseen visitors
new_visitors = pd.DataFrame({
    'minutes_on_site': [3.0, 10.0],
    'pages_visited': [2, 6]
})

# 4. Make the predictions
predictions = classifier.predict(new_visitors)

# 5. Peak under the hood at the actual probabilities
probabilities = classifier.predict_proba(new_visitors)

print(f"--- New Visitors DATAFRAME ---")
display(new_visitors)

print(f"Visitor A Prediction: {predictions[0]} (Probability of buying: {probabilities[0][1]*100:.1f}%)")
print(f"Visitor B Prediction: {predictions[1]} (Probability of buying: {probabilities[1][1]*100:.1f}%)")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
import numpy as np

# 1. Generate 100 rendom e-commerce visitors
np.random.seed(42)
minutes = np.random.uniform(1, 15, 100)
pages = np.random.randint(1, 10, 100)

# Simulate logic: Users who spend > 7 mins OR view > 5 pages tend to buy
bought = np.where((minutes > 7) | (pages > 5), 1, 0)

df_large_class = pd.DataFrame({'minutes_on_site': minutes, 'pages_visited': pages, 'bought_product': bought})
X_class = df_large_class[['minutes_on_site', 'pages_visited']]
y_class = df_large_class['bought_product']

# 2. The Train/Test Split (80% training, 20% testing)
X_train, X_test, y_train, y_test = train_test_split(X_class, y_class, test_size=0.2, random_state=42)

# 3. Train the model
model_eval = LogisticRegression()
model_eval.fit(X_train, y_train)

# 4. Predict the 20% vault data
test_preds = model_eval.predict(X_test)

# 5. Evaluate!
accuracy = accuracy_score(y_test, test_preds)
matrix = confusion_matrix(y_test, test_preds)

print(f"Overall Accuracy: {accuracy * 100:.1f}%\n")
print("--- Confusion Matrix ---")
print(matrix)
```
